### **1. STUDI KASUS GEOLISTRIK**

### **Deskripsi**

Kajian ini memakai keluaran survei **geolistrik**, seperti citra penampang resistivitas dua dimensi maupun data sounding.

Pembacaan data diarahkan untuk menemukan perubahan sifat listrik di bawah permukaan sehingga susunan lapisan, perbedaan material, struktur, atau zona tertentu dapat dikenali.

### **Role**

> Anda bertindak sebagai pendamping analisis yang membantu menerangkan informasi yang ditunjukkan oleh data geolistrik.

### **Task**

> Membantu membaca dan menafsirkan pola resistivitas yang tersedia sehingga perubahan karakter bawah permukaan dapat dijelaskan secara geofisika.

### **Context**

> Geolistrik bekerja dengan memanfaatkan perbedaan respons listrik material di bawah tanah. Resistivitas suatu material dapat berubah karena jenis material, porositas, kandungan air atau fluida, kejenuhan, dan lingkungan geologinya.

### **Input**

> Hasil yang digunakan berupa penampang resistivitas, angka resistivitas, informasi kedalaman, susunan atau posisi elektroda, serta keterangan geologi maupun hidrogeologi apabila tersedia.

### **Constraints**

> Tidak melakukan perhitungan ulang maupun menambahkan angka yang tidak terdapat pada data. Penjelasan harus bersumber dari informasi yang tersedia dan memisahkan fakta pengukuran, hasil penafsiran, serta kemungkinan interpretasi.

### **Output**

> Paparkan kecenderungan resistivitas, pembagian lapisan atau zona, perubahan arah mendatar dan vertikal, serta dugaan karakter material yang dapat menjelaskan pola tersebut.

### **Quality Criteria**

> Penafsiran harus mempertimbangkan legenda, rentang warna, besaran resistivitas, kedalaman, dan data pendukung. Istilah geofisika digunakan sesuai maknanya dan hasil dijelaskan secara runtut.

### **Verification**

> Tinjau kembali apakah uraian yang dibuat benar-benar ditunjukkan oleh penampang resistivitas. Jika ada data pendukung, lakukan pencocokan dengan informasi geologi, sumur atau bor, maupun hasil survei lainnya.


## Eksekusi 1: Geolistrik

**Data sintetis variasi 2:** lintasan 120 m, kedalaman model 24 m, zona resistivitas rendah pada jarak 28–58 m dan zona resistivitas tinggi di sisi kanan.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi, bukan data survei lapangan.

**Tujuan eksekusi:** menghasilkan keluaran terukur dan visual untuk contoh interpretasi.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# 1. PEMBUATAN MODEL RESISTIVITAS SINTETIS
x = np.linspace(0, 120, 121)       # Jarak lintasan (m)
z = np.linspace(0, 24, 49)         # Kedalaman (m)
X, Z = np.meshgrid(x, z)

rho = np.full_like(X, 75.0)        # Resistivitas latar belakang (ohm.m)
rho[(X >= 28) & (X <= 58) & (Z >= 5) & (Z <= 15)] = 22.0
rho[(X >= 82) & (X <= 108) & (Z >= 9)] = 185.0

print("=" * 55)
print("METODE GEOLISTRIK - MODEL RESISTIVITAS SINTETIS")
print("=" * 55)
print(f"Resistivitas minimum : {rho.min():.1f} ohm.m")
print(f"Resistivitas maksimum: {rho.max():.1f} ohm.m")
print("Panjang lintasan     : 120 m")
print("Kedalaman model      : 24 m")
print("=" * 55)

fig, ax = plt.subplots(figsize=(11, 5))
im = ax.pcolormesh(X, Z, rho, shading="auto", cmap="viridis", vmin=0, vmax=200)
ax.invert_yaxis()
fig.colorbar(im, ax=ax, pad=0.02).set_label("Resistivitas (ohm.m)")
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Penampang Resistivitas Sintetis - Variasi 2")
ax.set_xticks(np.arange(0, 121, 10))
ax.set_yticks(np.arange(0, 25, 3))
fig.tight_layout()
nama_file = "penampang_resistivitas_variasi2.png"
fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)
display(Image(filename=nama_file))
print(f"Grafik berhasil disimpan: {nama_file}")
```

Zona resistivitas rendah pada jarak sekitar 28–58 m dan kedalaman 5–15 m dapat berkaitan dengan material yang lebih konduktif, misalnya lempung atau material jenuh air. Zona resistivitas tinggi di sisi kanan menunjukkan kontras sifat listrik, tetapi penyebab geologinya tidak dapat dipastikan hanya dari model sintetis. Verifikasi memerlukan data geologi atau pengukuran pendukung.

## Eksekusi 2: Seismik Refraksi

**Data sintetis variasi 2:** 26 posisi penerima dengan jarak antarpenerima 6 m; waktu tiba dibuat dari model dua lapisan dengan kecepatan 650 m/s dan 2.100 m/s serta waktu intersep 0,038 s.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi, bukan data lapangan.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# 1. PEMBUATAN DATA SEISMIK REFRAKSI SINTETIS
x = np.arange(0, 151, 6, dtype=float)
v1 = 650.0      # Kecepatan lapisan 1 (m/s)
v2 = 2100.0     # Kecepatan lapisan 2 (m/s)
ti = 0.038      # Waktu intersep (s)

t_direct = x / v1
t_head = ti + x / v2
t = np.minimum(t_direct, t_head)
t_ms = t * 1000
t_direct_ms = t_direct * 1000
t_head_ms = t_head * 1000

print("=" * 55)
print("SEISMIK REFRAKSI - DATA SINTETIS VARIASI 2")
print("=" * 55)
print(f"Jumlah posisi penerima : {len(x)}")
print(f"Jarak antar penerima   : {x[1] - x[0]:.1f} m")
print(f"Kecepatan lapisan 1    : {v1:.1f} m/s")
print(f"Kecepatan lapisan 2    : {v2:.1f} m/s")
print(f"Waktu intersep         : {ti * 1000:.1f} ms")
print(f"Waktu tiba minimum     : {t_ms.min():.2f} ms")
print(f"Waktu tiba maksimum    : {t_ms.max():.2f} ms")
print("Parameter merupakan model sintetis.")
print("=" * 55)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, t_ms, "o-", linewidth=1.8, markersize=5, label="Waktu tiba sintetis")
ax.plot(x, t_direct_ms, "--", linewidth=1.5, label="Gelombang langsung")
ax.plot(x, t_head_ms, "--", linewidth=1.5, label="Gelombang refraksi")
ax.set_xlabel("Jarak sumber–penerima (m)")
ax.set_ylabel("Waktu tiba (ms)")
ax.set_title("Kurva Waktu Tiba Seismik Refraksi Sintetis - Variasi 2")
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
nama_file = "kurva_seismik_refraksi_variasi2.png"
fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)
display(Image(filename=nama_file))
print(f"Grafik berhasil disimpan: {nama_file}")
```

Kurva waktu tiba memperlihatkan perubahan kemiringan yang pada model refraksi sederhana berkaitan dengan perubahan kecepatan rambat. Model variasi ini menetapkan kecepatan lapisan bawah lebih tinggi daripada lapisan atas. Karena data dibangkitkan dari parameter yang telah ditentukan, hasilnya bukan estimasi kecepatan lapangan.
